anonymous function with single expression
syntax:lambda p1,p2:expression

```python
add_num=lambda n1,n2:n1+n2
print(add_num(100,200))
```

```python
sub_num=lambda n1,n2:n1-n2
print(sub_num(10,2))
```

```python
cube=lambda n1:n1**3
print(cube(3))
```

```python
max_two=lambda n1,n2:n1 if n1>n2 else n2
print(max_two(200,500))
```

```python
min_two=lambda n1,n2:n1 if n1<n2 else n2
print(min_two(200,500))
```

```python
user_experience= [
    ["hari",6],
    ["abin",1],
    ["bijoy",4],
    ["cibin",2],
    ["akash",7]
]
```

```python
#person with highest experience
print(max(user_experience,key=lambda lst:lst[1]))
```

```python
#person with lowest experience
print(min(user_experience,key=lambda lst:lst[1]))
```

```python
fr=open("movies.csv","r",encoding="utf-8")

movies=[]

for line in fr:
    line=line.rstrip("\n")
    id,title,language,year,run_time,rating,genre=line.split(",")
    movies_dict={
        "id":id,
        "title":title,
        "language":language,
        "year":year,
        "run_time":run_time,
        "rating":rating,
        "genre":genre

    }
    movies.append(movies_dict)
print(len(movies))
```

```python
all_titles=[m['title'] for m in movies]
all_titles
```

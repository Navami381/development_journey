movie case study

```python
movies=[
   ["kgf","kannada",150,2005,8],            #0
    ["balan","malayalam",130,2026,7],
    ["ramayan","hindi",150,2026,8],
    ["abcd","malayalam",140,2008,6],
    ["goatlyf","malayalam",160,2024,9]

]
```

```python
all_movietitles=[m[0] for m in movies]
all_movietitles
```

```python
all_lang={m[1] for m in movies}
all_lang
```

movie titles starting with a

```python
tilte_a=[m[0] for m in movies if m[0].startswith("a")]
tilte_a
```

```python
#display all malayalam movies

mala_movies=[m[0] for m in movies if m[1]=="malayalam"]
mala_movies
```

```python
#lang_wise count
languages=[m[1] for m in movies]
lc={l:languages.count(l) for l in languages}
lc
```

```python
import matplotlib.pyplot as plt

x=lc.keys()
y=lc.values()
plt.bar(x,y)
```

```python
#highest runtime movie
high_runtime = max([m[2] for m in movies])

high_runtime_movie = [m[0] for m in movies if m[2] == high_runtime]

high_runtime_movie


```

```python
#lowest runtime movie
low_runtime=min((m[2] for m in movies))
low_runtime_movie=[m[0] for m in movies if m[2]==low_runtime]
low_runtime_movie
```

```python
#avg runtime movie
avg_runtime=((m[2]) for m in movies)
average=sum(avg_runtime)/len(movies)
average
```

```python
#2026 released movies
movies=[m[0] for m in movies if m[-2]==2026]
movies
```
